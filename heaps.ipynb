```python
import heapq
import random
from typing import List
```

```python
def max_heapify(arr: List[int], n: int, index: int):
    largest = index
    left = 2 * largest + 1
    right = 2 * largest + 2

    if left < n and arr[largest] < arr[left]:
        largest = left
    
    if right < n and arr[largest] < arr[right]:
        largest = right
    
    if largest != index:
        arr[largest], arr[index] = arr[index], arr[largest]
        max_heapify(arr, n, largest)
        
def heap_sort(arr: List[int], n: int):
    for i in range(n // 2 - 1, -1, -1):
        max_heapify(arr, n, i)
    
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        max_heapify(arr, i, 0)
```

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        
        hashmap = {}
        for i in range(len(nums)):
            hashmap[nums[i]] = i
        
        result = []
        for i in range(len(nums)):

            number = nums[i]
            remaining = (target - number)

            if remaining in hashmap.keys():
                if hashmap[remaining] != i:
                    result.append(i)
                    result.append(hashmap[remaining])
                    break
        
        return result
```
